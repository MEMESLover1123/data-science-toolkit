```python
import pandas as pd

df = pd.read_csv('/content/sample_data/movie_blockbusters.csv')
display(df.head())
```

# 2. Evaluate if the data is good or not. Does it have enough samples?

```python
print("Rows and columns:", df.shape)
print("Repeated movie_id values:", df["movie_id"].duplicated().sum())
print("Blockbuster values (0 = No, 1 = Yes):")
print(df["blockbuster"].value_counts())
```

Yes, the dataset has enough data with 5,000 rows (movies) and 14 columns.

# 3. How many features are there? Is that enough?

```python
print("All columns:")
print(df.columns.tolist())
print("Number of columns:", len(df.columns))
print("Possible input columns (excluding blockbuster):", len(df.columns) - 1)
```

Yes, 13 potential input columns are provided.

# 4. How many samples/observation does it have? Is it enough?

```python
num_samples = len(df)
print("Total number of samples (rows):", num_samples)
print("Breakdown by target class (blockbuster):")
print(df['blockbuster'].value_counts())
print("\nPercentage breakdown:")
print(df['blockbuster'].value_counts(normalize=True) * 100)
```

Yes, 5000 observations is very enough.

# 5. What are the datatypes does the dataset consist of?

```python
print("Data types for each column:")
print(df.dtypes)
print("\nCount of columns per data type:")
print(df.dtypes.value_counts())
```

# 6. and 8. Which columns and rows have missing (null) values?

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())

# Rows with any missing values
missing_rows = df.isnull().any(axis=1).sum()
print("\nNumber of rows with missing values:", missing_rows)
```

# 7. Which are the features and which is the target?

```python
target = "blockbuster"
features = df.drop(columns=["blockbuster", "movie_id"])

print("Target:", target)
print("Example feature columns:", features.columns.tolist())
print("Target values:", df[target].unique())
```
