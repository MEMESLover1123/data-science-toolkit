```python
import pandas as pd

df =pd.read_csv('/content/drive/MyDrive/salary.csv')
display(df.head())
```

```python
# Check the column at index 6 and print it
print(df.columns[6])

# Rename the column at index 6 to 'Salary'
df.rename(columns={df.columns[6]: 'Salary'}, inplace=True)
display(df.head())
```

## 1. Incomplete data / Missing values

```python
# Find missing values (in pandas, NaN means Not a Number)
df.info()
```

```python
# Rename the column at index 7 to 'Compensation'
df.rename(columns={df.columns[7]: 'Compensation'}, inplace=True)
display(df.head())

# Question 1: Write code to count missing values in the Compensation column
print("Sum of null:", df['Compensation'].isna().sum())
```

```python
# Write a function that returns each column name with the number of missing values
def count_missing_values(df):
    print("--Columns and the number of missing values--")
    for column_name in df.columns:
        n_missing = df[column_name].isna().sum()
        print(f"{column_name} : {n_missing}")

# Run the function
count_missing_values(df)
```

### 1.1 Dropping columns with too many missing values

```python
# Sometimes if a column has too many missing values it may not be useful, so we can drop it
df.rename(columns={df.columns[5]: 'Context'}, inplace=True)
df.rename(columns={df.columns[9]: 'Other currency'}, inplace=True)
df.rename(columns={df.columns[10]: 'Salary context'}, inplace=True)

df = df.drop(
    columns=['Context', 'Other currency', 'Salary context']
)

df.head()
```

### 1.2 Dropping missing values

```python
# If you want to delete rows with missing values
df_clean = df.dropna()

# If you want to see the dropped rows:
# ~ this symbol will find the indexes not in df_clean
dropped_rows = df[~df.index.isin(df_clean.index)]

# df.index.isin(df_clean.index) returns True for rows that exist in both DataFrames.
# ~ flips those values (True becomes False), leaving you only with the rows that were removed.

print("Deleted / Dropped Rows:")
print(dropped_rows)

# NOTE: If we do dropna() without specifying any parameter, it will drop rows that have even 1 missing value.
```

```python
# Keep rows containing at least 2 non-missing values
df.dropna(thresh=2)

# Remove rows only if all their values are missing
df.dropna(how='all')

# Compare number of rows before and after dropna()
rows_before = len(df)
df_clean = df.dropna()
rows_after = len(df_clean)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

### 1.4 Impute missing values with most frequent

Imputation means filling in missing values instead of deleting rows.
You have three options:
* **Mean:** Fill using the average salary.
* **Median:** Fill using the middle value.
* **Most frequent:** Fill using the value appearing most often, which is 60000 here.

```python
# Question 5: Use SimpleImputer to fill missing values with the most frequent value.

# Import the library
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='most_frequent')

# Make a copy of your df
df_imputed = df.copy()

# Loop through every column in the DataFrame; if there is a null in any cell,
# fit_transform() learns which value appears most often and fills missing cells using that value.
for column in df_imputed.columns:
    if df_imputed[column].isna().any():
        df_imputed[[column]] = imputer.fit_transform(
            df_imputed[[column]]
        )

# Verify that your imputed columns now have 0 missing values
print(df_imputed.isna().sum())
df_imputed.head()
```

## 2. Noisy data
### 2.1 Remove outliers
* Noise can be random errors, outliers, or non-sensible data.
* Use boxplots or histograms (`df.boxplot()` or `df.hist()`) to find outliers.

```python
# Visualize outliers in the numeric columns (especially 'Salary')
df.boxplot(column=['Salary'])
```

```python
from scipy import stats
import numpy as np

# Method 1: Remove outliers manually (e.g., keeping salaries below 1,000,000)
df_no_outliers = df[df['Salary'] < 1000000]
print("Rows before outlier removal:", len(df))
print("Rows after manual outlier removal:", len(df_no_outliers))

# Method 2: Remove outliers using Z-score (keeping values within 3 standard deviations)
df_zscore = df[(np.abs(stats.zscore(df['Salary'])) < 3)]
print("Rows after Z-score outlier removal:", len(df_zscore))
```

### 2.2 Remove non-sensible values

```python
# Check summary statistics (look at min and max Salary values)
display(df.describe())

# Guide: df[df['col_name'] > threshold] to select sensible values
# Example: Keep only rows where Salary is greater than 1000 (removes $0 or $1 entries)
df_sensible = df[df['Salary'] > 1000]
display(df_sensible.describe())
```

## 3. Inconsistent data
* Find 'duplicated' categories or inconsistent entries (for example, `'Academia'`, `'Academic'`, and `'Academia/Research'` should be the same).
* Find typos and redundant categories.

```python
# Rename 'What is your gender?' to 'Gender' so it is easier to work with
if 'What is your gender?' in df.columns:
    df.rename(columns={'What is your gender?': 'Gender'}, inplace=True)

# Check the unique items for Gender
# Guide: df['Gender'].unique()
print("Unique values in Gender before cleaning:")
print(df['Gender'].unique())
```

```python
# Remove redundancy by grouping 'Other or prefer not to answer' and 'Prefer not to answer' into 'Other'
# Guide: df.replace()
df['Gender'] = df['Gender'].replace(
    ['Other or prefer not to answer', 'Prefer not to answer'],
    'Other'
)
```

```python
# Check whether the values have been replaced or not
# Guide: df['Gender'].unique()
print("Unique values in Gender after cleaning:")
print(df['Gender'].unique())
```
