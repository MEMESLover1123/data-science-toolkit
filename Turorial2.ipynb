```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following
#1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([
    [1, 2],
    [3, 4]
])

print("This is my numpy 2*2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4
random3x4 = np.random.rand(3,4)
print("Random 3*4 array")
print(random3x4)

```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of ones:")
print(ones_array)

print("Array of zeroes:")
print(zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
random_6x3 = np.random.rand(6, 3)
transposed_array = random_6x3.T

print("Original 6x3 array:")
print(random_6x3)

print("\nTransposed 3x6 array:")
print(transposed_array)
```

```python
# 5. Create two random numpy array of size n x p and p x m and then compute the dot product
n, p, m = 3, 4, 2

array_np = np.random.rand(n, p)
array_pm = np.random.rand(p, m)
dot_product = np.dot(array_np, array_pm)

print(f"Array A (size {n}x{p}):")
print(array_np)

print(f"\nArray B (size {p}x{m}):")
print(array_pm)

print("\nDot Product A * B:")
print(dot_product)
```
